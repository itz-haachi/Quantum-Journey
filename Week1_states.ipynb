Week 1 – Quantum states in NumPy: kets, norms, inner products, the Born rule, and simulated measurement

```python
print("Hello Quantum")
```

```python
import numpy as np
ket0 = np.array([1,0],dtype=complex)
ket1 = np.array([0,1],dtype=complex)
plus = (ket0 + ket1)/np.sqrt(2)
minus = (ket0 - ket1)/np.sqrt(2)
print(plus)
```

```python
def norm(v):
  return np.sqrt(np.vdot(v,v).real)
print(norm(ket0),norm(plus),norm(np.array([3,4j])))
```

```python
u = np.array([1,1j])
v = np.array([1,-1j])
print("np.dot :", np.dot(u, v))
print("np.vdot:", np.vdot(u, v))
```

```python
def prob(outcome,state):
  return abs(np.vdot(outcome,state)) ** 2
psi = np.array([1, 1j]) / np.sqrt(2)
print(prob(ket0, psi), prob(ket1, psi), prob(plus, ket0))
```

```python
def normalise(v) :
  return v / norm(v)
print(normalise(np.array([3, 4j])))
```

```python
rng = np.random.default_rng(42)
p0 = prob(ket0, plus)
p1 = prob(ket1, plus)
samples = rng.choice([0,1],size=1000,p=[p0,p1])
print(np.bincount(samples))
```

```python
import matplotlib.pyplot as plt
plt.bar(["0", "1"], np.bincount(samples))
plt.title("1000 measurements of |+>")
plt.show()
```
